```python
"""
Generates a sample CSV to test the Streamlit app's "Batch Forecast" tab.
Run this AFTER eda_and_modeling.py has run at least once.
"""

import joblib
import pandas as pd

feature_cols = joblib.load("outputs/feature_columns.pkl")
history = joblib.load("outputs/recent_history.pkl")

# Take a random sample of 20 real historical rows (already has every
# engineered feature the model expects) and drop the target column.
sample = history.sample(20, random_state=42)[feature_cols]

sample.to_csv("outputs/sample_batch_forecast.csv", index=False)
print("Saved outputs/sample_batch_forecast.csv —", len(sample), "rows")
```
